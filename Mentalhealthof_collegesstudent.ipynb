```python
df['mental_health_score'] = df[['depression','anxiety','isolation','future_insecurity']].sum(axis=1)


def categorize_cgpa(x):
    if x < 2.5:
        return 'Low'
    elif 2.5 <= x <= 3.0:
        return 'Medium'
    else:
        return 'High'
df['cgpa_numeric'] = df['cgpa'].apply(lambda x: (float(x.split('-')[0]) + float(x.split('-')[1])) / 2 if '-' in str(x) else float(x))

df['cgpa_category'] = df['cgpa_numeric'].apply(categorize_cgpa)

df.head()
```

```python


print("Average Mental Health Score by Sports Engagement:")
print(df.groupby('sports_engagement')['mental_health_score'].mean())

print("\nStudy Satisfaction by CGPA Category:")
print(df.groupby('cgpa_category')['study_satisfaction'].mean())

print("\nFuture Insecurity by Residential Status:")
print(df.groupby('residential_status')['future_insecurity'].mean())
```

```python
cols = ['cgpa_numeric', 'academic_workload ', 'academic_pressure',
        'study_satisfaction', 'depression', 'anxiety', 'isolation', 'future_insecurity']

plt.figure(figsize=(8,6))
sns.heatmap(df[cols].corr(), annot=True, fmt=".2f", cmap="viridis", cbar=True)
plt.title("Focused Heatmap: Academic vs Mental Health", fontsize=14)
plt.show()
```

```python
cols = ['cgpa_numeric', 'academic_workload ', 'academic_pressure',
        'study_satisfaction', 'depression', 'anxiety', 'isolation', 'future_insecurity']

corr_matrix = df[cols].corr()

plt.figure(figsize=(10,6))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # upper triangle mask
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", mask=mask)
plt.title("Heatmap (Lower Triangle Only)", fontsize=14)
plt.show()
```

```python
fig=plt.figure (figsize=(6,2))
ax1 =fig.add_axes([0,0,1,1])
#ax2 =fig.add_axes([0,0.5,1,1])
ax1.plot(x,y,color="red", marker ='o',markersize = 5,linestyle ='--')
ax1.plot(x,z ,color="blue", marker ='o',markersize = 5,linestyle ='--')
ax1.legend ()
ax1.set_xlabel('Angle')
ax1.set_ylabel('sins/ co of angle ')
ax1.set_title ('sine/ co function' )
```

```python
x = np.arange(0, 2*np.pi, 0.5)

y = np.sin(x)

z= np.cos(x)

a=np.tan(x)

b = np.log10(x)

fig= plt.figure(figsize = (6,4))

ax1 =fig.add_axes ([0,0,0.45,0.45])

ax2 =fig.add_axes ([0.55,0,0.45,0.45])

ax3 =fig.add_axes ([0,0.55,0.45,0.45])

ax4 =fig.add_axes ([0.55,0.55,0.45,0.45])

ax1.plot(x, y, color='blue', marker='o', markersize=5, linestyle='--', label = 'Sine')
ax1.legend(loc='lower left')
ax2.plot(x,z, color='red', marker='o', markersize=5, linestyle='--', label= 'cos')
ax2.legend ()
ax3.plot(x,a, color = 'green', marker = 'o', markersize=5, linestyle='--', label='tan')
ax3.legend ()
ax4.plot(x,b, color='black', marker='o', markersize = 5, linestyle='--', label= 'log')
ax4.legend ()
```
